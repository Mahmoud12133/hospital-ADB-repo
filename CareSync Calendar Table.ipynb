```python
%sql
CREATE OR REPLACE TABLE caresyncadb.silver.calendar AS
WITH dates AS (
  SELECT explode(sequence(
    DATE '1940-01-01',
    DATE '2030-12-31',
    INTERVAL 1 DAY
  )) AS calendar_date
)
SELECT
  calendar_date,
  YEAR(calendar_date)                          AS year,
  MONTH(calendar_date)                         AS month,
  DAY(calendar_date)                           AS day,
  QUARTER(calendar_date)                       AS quarter,
  DAYOFWEEK(calendar_date)                     AS day_of_week,
  DAYOFYEAR(calendar_date)                     AS day_of_year,
  WEEKOFYEAR(calendar_date)                    AS week_of_year,
  DATE_FORMAT(calendar_date, 'EEEE')           AS day_name,
  DATE_FORMAT(calendar_date, 'MMMM')          AS month_name,
  CASE
    WHEN DAYOFWEEK(calendar_date) IN (1, 7) THEN TRUE
    ELSE FALSE
  END                                          AS is_weekend,
  CASE
    WHEN (YEAR(calendar_date) % 4 = 0 AND YEAR(calendar_date) % 100 != 0)
      OR YEAR(calendar_date) % 400 = 0 THEN TRUE
    ELSE FALSE
  END                                          AS is_leap_year,
  TRUNC(calendar_date, 'MM')                   AS first_day_of_month,
  LAST_DAY(calendar_date)                      AS last_day_of_month,
  TRUNC(calendar_date, 'QUARTER')              AS first_day_of_quarter,
  LAST_DAY(ADD_MONTHS(TRUNC(calendar_date, 'QUARTER'), 2)) AS last_day_of_quarter,
  TRUNC(calendar_date, 'YEAR')                 AS first_day_of_year,
  DATE_FORMAT(calendar_date, 'yyyyMMdd')       AS date_key_int
FROM dates
```

```python
%sql
SELECT COUNT(*) AS total_rows, MIN(calendar_date) AS min_date, MAX(calendar_date) AS max_date
FROM caresyncadb.silver.calendar
```
